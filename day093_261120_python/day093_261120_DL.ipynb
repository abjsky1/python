```python
# 파이토치는 텐서(tensor) 객체 사용
# 파이토치 라이브러리 호출
import torch
import numpy as np
np_data = np.array( [ [ 10.0 , 20.0 , 30.0 ] , [ 40.0 , 50.0 , 60.0 ] ] )
# torch.tensor( 넘파이배열 , dtype=torch.타입명 )
x = torch.tensor( np_data , dtype=torch.float32 )
print(x)
print(x.shape , x.dtype)

# 차원 재구성
# - 차원 변경
y = x.reshape(3,2)
print(y)
print(y.shape)
# - 차원추가(인덱스0자리에)
y = x.unsqueeze(0)
print(y)
print(y.shape)
# - 차원추가(인덱스-1자리에)
y = x.unsqueeze(-1)
print(y)
print(y.shape)
```

```python
import torch
sample = torch.tensor( [ -2.0 , 0.0 , 3.0 ] )
print(sample)

# 시그모이드 활성화 함수
# 0 ~ 1 사이의 확률값으로 변환하는 함수
# ( 1 / ( 1 + e^-z ) )
print( torch.sigmoid(sample) ) 

# ReLU 활성화함수
# 음수는 0 으로 차단하고 양수만 표현하는 함수
# max( 0 ,z )
print( torch.relu(sample) )
```

```python
import torch
# x 에 특성값 설정
x = torch.tensor( 2.0 ) 
# y 에 타겟값 설정
y = torch.tensor( 1.0 )
# w 에 가중치 설정
# requires_grad=True : 기울기 기록 (역전파 기록)
w = torch.tensor( 0.2 , requires_grad=True )
# b 에 편향값 설정
b = torch.tensor( 0.0 , requires_grad=True )
# 학습률 설정
learning_rate = 0.1
# 가중치 계산
# y = wx + b
y_pred = w * x + b
print( y_pred )
# 활성화함수 
# 양수면 그대로 , 음수면 0
y_pred = torch.relu( y_pred ) 
print( y_pred )
# 손실 함수 라이브러리 호출
import torch.nn as nn
# 손실 함수 (오차 구해주는 함수)
# MSE : mean squere error
criterion = nn.MSELoss()
loss = criterion(y_pred , y)
print(loss)
# 역전파
loss.backward()
print(w.grad.item() , b.grad.item())
# 기울기 기록을 일시 중지 블록 선언
with torch.no_grad():
  # 가중치 업데이트
  # 경사하강법 : 가중치 = 기존가중치 - ( 학습률 * 기울기 )
  w -= learning_rate * w.grad.item()
  b -= learning_rate * b.grad.item()
  print(w)
  print(b)
  # 가중치/편향 기울기 0으로 초기화
  w.grad.zero_()
  b.grad.zero_() 
print('한번 업데이트 이후 :' , w.item() , b.item() )
print( w * x + b )
```
