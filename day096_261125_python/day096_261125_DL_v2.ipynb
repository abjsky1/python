```python
from inspect import classify_class_attrs
# 구글에서 코랩 업로드
from google.colab import files
upload = files.upload()
image_path = list(upload.keys())
print(image_path)

# 사전학습된 ResNet50 모델 사용
import torch
from torchvision import models
# ResNet50 모델의 가중치 기본값
weights = models.ResNet50_Weights.DEFAULT
# 해당 모델의 기본값 가중치 대입
model = models.resnet50( weights = weights )
# 이미지 전처리 : 흑백-1차원 , png(rgba)-4차원
# 전처리 파이프라인 제공
preprocess = weights.transforms()
# 이미치 처리 제공 라이브러리
from PIL import Image
# 강제로 컬러 이미지 변경
# (3 , 가로 , 세로)
img = Image.open( image_path[0] ).convert('RGB')
# 미니배치 위한 4차원 변경
# (배치 , 3 , 가로 , 세로)
input_tensor = preprocess(img).unsqueeze(0)


model.eval() # 모델 학습이 아닌 추론/평가모드
with torch.no_grad() :
    # 추론
    pred = model(input_tensor)
    # 다중분류(1000개)
    preds = torch.nn.functional.softmax(pred[0], dim=0)

# 확률이 큰 5개 클래스 추출
# torch.topk( 확률분호 , k )
top5_pred , top5_index = torch.topk( preds , 5 )
for i in range(5) : 
    # 사전모델 가중치에서 메타정보 클래스 속성명
    class_name = weights.meta['categories'][top5_index[i].item()]
    print(class_name)
    print(top5_pred[i].item()*100)


# 사전학습에 반영된 클래스/정답지만 추론 가능
# 그외 불가능 하므로 전이학습 필요
```
